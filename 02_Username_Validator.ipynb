```python
# Must not be greater than 12 characters
# Must not contain any spaces
# MUST not contain any digits



username = input("Enter a username: ")

if len(username) > 12:
  print(f"Your username cannot be greater than 12 characters")

elif not username.find(" ") == -1:
  print(f"Your username cannot contain spaces")

elif not username.isalpha():
  print(f"Your username cannot contain digits")

else:
  print(f"Your username is valid")
```
