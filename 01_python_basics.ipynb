```python
# Logical operators = evaluate multiple conditionsv (or, and ,not)
#                   or = at least one condition must be true
#                  and = both conditions must be True
#                  not = inverts the conditions (not False, not True)
```

```python
#OR

temp  = 23
is_raining = False

if temp > 34 or temp < 0 or is_raining:
  print("The outdoor event is cancelled")
else:
  print("The outdoor event is still scheduled")
```

```python
temp = 21
is_sunny = True

if temp >= 28 and is_sunny:
  print("It is HOT outside!!")
  print("It is SUNNY!")
elif temp <0 and is_sunny:
  print("It is COLD outside!!")
  print("It is SUNNY!")
elif 28 > temp > 0 and is_sunny:
  print("It is WARM outside!!")
  print("It is SUNNY!")
```

```python
#NOT

temp = 26
is_sunny = False

if temp >= 28 and is_sunny:
  print("It is HOT outside!!")
  print("It is SUNNY!")
elif temp <0 and is_sunny:
  print("It is COLD outside!!")
  print("It is SUNNY!")
elif 28 > temp > 0 and is_sunny:
  print("It is WARM outside!!")
  print("It is SUNNY!")
elif temp >= 28 and not is_sunny:
  print("It is HOT outside!!")
  print("It is CLOUDY!")
elif temp <0 and not is_sunny:
  print("It is COLD outside!!")
  print("It is CLOUDY!")
elif 28 > temp > 0 and not is_sunny:
  print("It is WARM outside!!")
  print("It is CLOUDY!")
```

FORMAT SPECIFIERS IN THE FORM OF "f" STRINGS

```python
# format specifiers = {value:flags} format a value based on what
#                              flags are inserted
```

```python
# Decimal point precision

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:.2f}")
print(f"Price 2 is R{price2:.2f}")
print(f"Price 3 is R{price3:.2f}")
```

```python
#Left justifiction

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:<10}")
print(f"Price 2 is R{price2:<10}")
print(f"Price 3 is R{price3:<10}")
```

```python
# Right Justification

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:>10}")
print(f"Price 2 is R{price2:>10}")
print(f"Price 3 is R{price3:>10}")
```

```python
# Centre align

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:^10}")
print(f"Price 2 is R{price2:^10}")
print(f"Price 3 is R{price3:^10}")
```

```python
# Showing that a number is positive (+)

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:+}")
print(f"Price 2 is R{price2:+}")
print(f"Price 3 is R{price3:+}")
```

```python
# Showing that a number is positive (use space)

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1: }")
print(f"Price 2 is R{price2: }")
print(f"Price 3 is R{price3: }")
```

```python
# Thousand seperator (use a comma)

price1 = 3000.14159
price2 = -9873.65
price3 = 1232.34

print(f"Price 1 is R{price1:,}")
print(f"Price 2 is R{price2:,}")
print(f"Price 3 is R{price3:,}")
```

```python
# Mix

price1 = 3000.14159
price2 = -9873.65
price3 = 1232.34

print(f"Price 1 is R{price1:+,.2f}")
print(f"Price 2 is R{price2:+,.2f}")
print(f"Price 3 is R{price3:+,.2f}")
```

WHILE LOOPS

```python
# While loop = execut some code WHILE some condition remains true

# example

name = input("Enter your name name: ")

while name == "":
  print("You did not enter your name!")
  name = input("Enter your name name: ")

print(f"Hello {name}")
```

```python
age = int(input("Enter your age: "))

while age < 0:
  print("Age can't be negative!")
  age = int(input("Enter your age: "))

print(f"Your {age} years old!")
```

```python
# introduction of logical operators

food = input("Enter a food you like (q to quit): ")

while not food == "q":
  print(f"You like {food}")
  food = input("Enter a food you like (q to quit): ")

print("bye")
```

```python
# OR logical operator

num = int(input("Enter a # between 1 - 10: "))

while num < 1 or num > 10:
  print(f"{num} is not valid!")
  num = int(input("Enter a # between 1 - 10: "))

print(f"Your number is {num}")
```
