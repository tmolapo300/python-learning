FORMAT SPECIFIERS IN THE FORM OF "f" STRINGS

```python
# format specifiers = {value:flags} format a value based on what
#                              flags are inserted
```

```python
# Decimal point precision

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:.2f}")
print(f"Price 2 is R{price2:.2f}")
print(f"Price 3 is R{price3:.2f}")
```

```python
#Left justifiction

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:<10}")
print(f"Price 2 is R{price2:<10}")
print(f"Price 3 is R{price3:<10}")
```

```python
# Right Justification

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:>10}")
print(f"Price 2 is R{price2:>10}")
print(f"Price 3 is R{price3:>10}")
```

```python
# Centre align

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:^10}")
print(f"Price 2 is R{price2:^10}")
print(f"Price 3 is R{price3:^10}")
```

```python
# Showing that a number is positive (+)

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1:+}")
print(f"Price 2 is R{price2:+}")
print(f"Price 3 is R{price3:+}")
```

```python
# Showing that a number is positive (use space)

price1 = 3.14159
price2 = -987.65
price3 = 12.34

print(f"Price 1 is R{price1: }")
print(f"Price 2 is R{price2: }")
print(f"Price 3 is R{price3: }")
```

```python
# Thousand seperator (use a comma)

price1 = 3000.14159
price2 = -9873.65
price3 = 1232.34

print(f"Price 1 is R{price1:,}")
print(f"Price 2 is R{price2:,}")
print(f"Price 3 is R{price3:,}")
```

```python
# Mix

price1 = 3000.14159
price2 = -9873.65
price3 = 1232.34

print(f"Price 1 is R{price1:+,.2f}")
print(f"Price 2 is R{price2:+,.2f}")
print(f"Price 3 is R{price3:+,.2f}")
```

```python
questions = ("What is the first letter of the Roman alphabet?: ",
             "What is the colour of the sky?: ",
             "What is the capital of England?: ",
             "Where is the city of Budapest located?: ")

options = (("A. X ","B. S ","C. O ","D. A "),
           ("A. Brown ","B. Pink ","C. Indigo ","D. Blue "),
           ("A. Brussels ","B. Bloemfontein ","C. Cardiff ","D. London "),
           ("A. Belgium ","B. Serbia ","C. Russia ","D. Hungary "))

answers = ("D", "D", "D", "D")

guesses = []
score = 0
question_num = 0

for question in questions:
  print("----------------")
  print(question)
  for option in options[question_num]:
    print(option)

  guess = input("Enter (A, B, C, D): ").upper()
  guesses.append(guess)
  if guess == answers[question_num]:
    score += 1
    print("CORRECT!")
  else:
    print("INCORRECT!")
    print(f"{answers[question_num]} is the correct answer")

  question_num += 1

# Move results printing and final score calculation outside the loop
print("-------------------")
print("      RESULTS      ")
print("-------------------")

print("answers: ", end="")
for answer in answers:
  print(answer, end=" ")
print()

print("guesses: ", end="")
for guess in guesses:
  print(guess, end=" ")
print()

final_score_percentage = int(score / len(questions) * 100)
print(f"Your score is: {final_score_percentage}%")
```

```python
# Removal

capitals = {"Russia":"Moscow",
            "France":"Paris",
            "UK":"London",
            "China":"Beijing"}

capitals.pop("China")

print(capitals)
```
