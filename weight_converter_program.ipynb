```python
#Weight converter
```

```python
weight = float(input("Enter your weight: "))
unit = input("(L)bs or (K)g?: ")

if unit == "K":
  weight = weight * 2.205
  unit = "Lbs."
  print(f"Your weight is: {round(weight, 2)} {unit}")
elif unit == "L":
  weight = weight / 2.205
  unit = "Kgs."
  print(f"Your weight is: {round(weight, 2)} {unit}")
else:
  print(f"{unit} is not a valid unit")
```
