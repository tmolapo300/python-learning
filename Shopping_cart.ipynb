```python
# Multi-item shopping cart

foods = []
prices = []

while True:
  food = input("Enter a food you want to purchase (q to quit): ")
  if food.lower() == "q":
      break
  else:
    price = float(input("Enter the price of the food: "))
    foods.append(food)
    prices.append(price)

print("\n--- YOUR CART ---")
total = 0

for i in range(len(foods)):
  print(f"{foods[i]}  R{prices[i]:.2f}")
  total += prices[i]

print(f"\nTotal Bill: R{total:.2f}")
```
